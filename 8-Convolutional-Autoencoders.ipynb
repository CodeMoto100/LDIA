```python
# ================================================================
# AUTOENCODER FOR IMAGE COMPRESSION AND RECONSTRUCTION
# Complete Google Colab - SINGLE CELL
# Dataset: MNIST
# ================================================================

# -------------------- 1. IMPORT LIBRARIES -----------------------

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

from skimage.metrics import structural_similarity as ssim
from skimage.metrics import peak_signal_noise_ratio as psnr

print("TensorFlow Version:", tf.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))


# -------------------- 2. LOAD DATASET ---------------------------

(x_train, _), (x_test, _) = mnist.load_data()

# Normalize pixels to [0,1]
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Add channel dimension
x_train = np.expand_dims(x_train, axis=-1)
x_test = np.expand_dims(x_test, axis=-1)

print("\nTraining data:", x_train.shape)
print("Testing data :", x_test.shape)


# -------------------- 3. BUILD AUTOENCODER ----------------------

input_img = layers.Input(
    shape=(28, 28, 1),
    name="input_image"
)

# ==================== ENCODER ====================

x = layers.Conv2D(
    32, (3, 3),
    activation="relu",
    padding="same"
)(input_img)

x = layers.BatchNormalization()(x)

x = layers.MaxPooling2D(
    (2, 2),
    padding="same"
)(x)
# 28 x 28 x 32 -> 14 x 14 x 32

x = layers.Conv2D(
    64, (3, 3),
    activation="relu",
    padding="same"
)(x)

x = layers.BatchNormalization()(x)

x = layers.MaxPooling2D(
    (2, 2),
    padding="same"
)(x)
# 14 x 14 x 64 -> 7 x 7 x 64

# -------- Compressed Latent Representation --------

latent = layers.Conv2D(
    4, (3, 3),
    activation="relu",
    padding="same",
    name="latent_space"
)(x)
# 7 x 7 x 4 = 196 values


# ==================== DECODER ====================

x = layers.Conv2D(
    64, (3, 3),
    activation="relu",
    padding="same"
)(latent)

x = layers.UpSampling2D(
    (2, 2)
)(x)
# 7 x 7 -> 14 x 14

x = layers.Conv2D(
    32, (3, 3),
    activation="relu",
    padding="same"
)(x)

x = layers.UpSampling2D(
    (2, 2)
)(x)
# 14 x 14 -> 28 x 28

decoded = layers.Conv2D(
    1, (3, 3),
    activation="sigmoid",
    padding="same",
    name="reconstructed_image"
)(x)


# Complete autoencoder
autoencoder = models.Model(
    input_img,
    decoded,
    name="MNIST_Autoencoder"
)

print("\n================ MODEL SUMMARY ================")
autoencoder.summary()


# -------------------- 4. COMPILE MODEL --------------------------

autoencoder.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="binary_crossentropy"
)


# -------------------- 5. TRAIN MODEL ----------------------------

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=2,
    min_lr=1e-6
)

print("\n================ TRAINING ================")

history = autoencoder.fit(
    x_train,
    x_train,
    epochs=30,
    batch_size=128,
    shuffle=True,
    validation_data=(x_test, x_test),
    callbacks=[early_stop, reduce_lr],
    verbose=1
)


# -------------------- 6. PLOT TRAINING LOSS ---------------------

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss",
    linewidth=2
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss",
    linewidth=2
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Autoencoder Training and Validation Loss")
plt.legend()
plt.grid(alpha=0.3)

plt.show()


# -------------------- 7. RECONSTRUCT IMAGES --------------------

print("\n================ RECONSTRUCTION ================")

reconstructed = autoencoder.predict(
    x_test,
    batch_size=128,
    verbose=1
)

print("Original shape     :", x_test.shape)
print("Reconstructed shape:", reconstructed.shape)

print("\nReconstructed pixel range:")
print("Minimum:", reconstructed.min())
print("Maximum:", reconstructed.max())


# -------------------- 8. DISPLAY RESULTS ------------------------

n = 10

plt.figure(figsize=(20, 5))

# Original images
for i in range(n):

    ax = plt.subplot(2, n, i + 1)

    plt.imshow(
        x_test[i].squeeze(),
        cmap="gray",
        vmin=0,
        vmax=1
    )

    plt.axis("off")

    if i == 0:
        ax.set_title("Original")


# Reconstructed images
for i in range(n):

    ax = plt.subplot(
        2, n, i + 1 + n
    )

    plt.imshow(
        reconstructed[i].squeeze(),
        cmap="gray",
        vmin=0,
        vmax=1
    )

    plt.axis("off")

    if i == 0:
        ax.set_title("Reconstructed")


plt.suptitle(
    "Original vs Reconstructed Images",
    fontsize=16
)

plt.tight_layout()
plt.show()


# -------------------- 9. CALCULATE MSE --------------------------

mse_values = np.mean(
    np.square(
        x_test - reconstructed
    ),
    axis=(1, 2, 3)
)

mean_mse = np.mean(mse_values)


# -------------------- 10. CALCULATE PSNR & SSIM -----------------

psnr_values = []
ssim_values = []

for i in range(len(x_test)):

    original = x_test[i].squeeze()
    recon = reconstructed[i].squeeze()

    psnr_values.append(
        psnr(
            original,
            recon,
            data_range=1.0
        )
    )

    ssim_values.append(
        ssim(
            original,
            recon,
            data_range=1.0
        )
    )

mean_psnr = np.mean(psnr_values)
mean_ssim = np.mean(ssim_values)


# -------------------- 11. ENCODER -------------------------------

encoder = models.Model(
    input_img,
    latent,
    name="Encoder"
)

latent_test = encoder.predict(
    x_test,
    batch_size=128,
    verbose=0
)

print("\n================ LATENT REPRESENTATION ================")

print(
    "Original image shape :",
    x_test.shape[1:]
)

print(
    "Latent shape         :",
    latent_test.shape[1:]
)


# -------------------- 12. COMPRESSION ANALYSIS -----------------

original_elements = 28 * 28 * 1

latent_elements = (
    latent_test.shape[1] *
    latent_test.shape[2] *
    latent_test.shape[3]
)

compression_ratio = (
    original_elements /
    latent_elements
)

size_reduction = (
    1 -
    latent_elements / original_elements
) * 100


# -------------------- 13. FINAL RESULTS -------------------------

print("\n")
print("=" * 60)
print("           FINAL AUTOENCODER RESULTS")
print("=" * 60)

print(f"Reconstruction MSE       : {mean_mse:.6f}")
print(f"Reconstruction PSNR      : {mean_psnr:.2f} dB")
print(f"Reconstruction SSIM      : {mean_ssim:.4f}")

print("\n--- Compression ---")

print(
    f"Original representation : "
    f"28 x 28 x 1 = {original_elements} values"
)

print(
    f"Latent representation   : "
    f"7 x 7 x 4 = {latent_elements} values"
)

print(
    f"Compression Ratio       : "
    f"{compression_ratio:.2f}:1"
)

print(
    f"Size Reduction          : "
    f"{size_reduction:.2f}%"
)

print("=" * 60)


# -------------------- 14. METRIC DISTRIBUTIONS -----------------

plt.figure(figsize=(15, 4))

# MSE
plt.subplot(1, 3, 1)

plt.hist(
    mse_values,
    bins=30,
    color="blue",
    alpha=0.7
)

plt.xlabel("MSE")
plt.ylabel("Number of Images")
plt.title("MSE Distribution")


# PSNR
plt.subplot(1, 3, 2)

plt.hist(
    psnr_values,
    bins=30,
    color="green",
    alpha=0.7
)

plt.xlabel("PSNR (dB)")
plt.ylabel("Number of Images")
plt.title("PSNR Distribution")


# SSIM
plt.subplot(1, 3, 3)

plt.hist(
    ssim_values,
    bins=30,
    color="orange",
    alpha=0.7
)

plt.xlabel("SSIM")
plt.ylabel("Number of Images")
plt.title("SSIM Distribution")

plt.tight_layout()
plt.show()


# -------------------- 15. LATENT FEATURES ----------------------

plt.figure(figsize=(10, 8))

for i in range(4):

    plt.subplot(2, 2, i + 1)

    plt.imshow(
        latent_test[0, :, :, i],
        cmap="viridis"
    )

    plt.title(
        f"Latent Feature {i + 1}"
    )

    plt.axis("off")

plt.suptitle(
    "Latent Representation Feature Maps"
)

plt.tight_layout()
plt.show()


# -------------------- 16. SAVE MODEL ----------------------------

autoencoder.save(
    "mnist_autoencoder.keras"
)

encoder.save(
    "mnist_encoder.keras"
)

print("\nModels saved successfully:")
print("  mnist_autoencoder.keras")
print("  mnist_encoder.keras")

print("\n================ EXPERIMENT COMPLETED ================")
```
