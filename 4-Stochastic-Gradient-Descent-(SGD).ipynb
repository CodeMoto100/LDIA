```python
import math


# ============================================================
# SIGMOID ACTIVATION FUNCTION
# ============================================================

def sigmoid(x):
    return 1 / (1 + math.exp(-x))


# ============================================================
# SINGLE LAYER FEED-FORWARD NEURAL NETWORK
# STOCHASTIC GRADIENT DESCENT
# ============================================================

class SingleLayerNN:

    def __init__(self, n_inputs, learning_rate=0.1):
        self.weights = [0.0] * n_inputs
        self.bias = 0.0
        self.learning_rate = learning_rate

    # --------------------------------------------------------
    # Forward propagation
    # --------------------------------------------------------

    def predict(self, x):

        wx = sum(w * xi for w, xi in zip(self.weights, x))
        z = wx + self.bias

        y_pred = sigmoid(z)

        return wx, z, y_pred

    # --------------------------------------------------------
    # Training using Stochastic Gradient Descent
    # --------------------------------------------------------

    def train(self, X, y, epochs):

        for epoch in range(epochs):

            for xi, target in zip(X, y):

                # Forward propagation
                wx, z, prediction = self.predict(xi)

                # Error
                error = prediction - target

                # SGD weight update
                for i in range(len(self.weights)):

                    self.weights[i] -= (
                        self.learning_rate
                        * error
                        * xi[i]
                    )

                # Bias update
                self.bias -= (
                    self.learning_rate * error
                )


# ============================================================
# AND GATE DATASET
# ============================================================

X = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

y = [
    0,
    0,
    0,
    1
]


# ============================================================
# CREATE MODEL
# ============================================================

model = SingleLayerNN(
    n_inputs=2,
    learning_rate=0.1
)


# ============================================================
# INITIAL PARAMETERS
# ============================================================

print("\n" + "=" * 75)
print("       SINGLE-LAYER FEED-FORWARD NEURAL NETWORK")
print("       STOCHASTIC GRADIENT DESCENT")
print("=" * 75)

print("\nInitial Parameters:")
print("-" * 75)
print(f"Initial W1   : {model.weights[0]:.4f}")
print(f"Initial W2   : {model.weights[1]:.4f}")
print(f"Initial Bias : {model.bias:.4f}")

print("\nTraining Parameters:")
print("-" * 75)
print(f"Learning Rate : {model.learning_rate}")
print(f"Epochs        : 10000")


# ============================================================
# TRAIN MODEL
# ============================================================

model.train(
    X,
    y,
    epochs=10000
)


# ============================================================
# FINAL PARAMETERS
# ============================================================

print("\n" + "=" * 75)
print("                     FINAL PARAMETERS")
print("=" * 75)

print(f"Final W1   : {model.weights[0]:.4f}")
print(f"Final W2   : {model.weights[1]:.4f}")
print(f"Final Bias : {model.bias:.4f}")


# ============================================================
# FINAL PREDICTIONS
# ============================================================

print("\n" + "=" * 85)
print("                       FINAL PREDICTIONS")
print("=" * 85)

print(
    f"{'X1':^8} | {'X2':^8} | "
    f"{'Target':^10} | {'WX':^10} | "
    f"{'WX + Bias':^12} | {'Prediction':^12} | {'Class':^8}"
)

print("-" * 85)

for xi, target in zip(X, y):

    wx, z, prediction = model.predict(xi)

    predicted_class = 1 if prediction >= 0.5 else 0

    print(
        f"{xi[0]:^8} | {xi[1]:^8} | "
        f"{target:^10} | {wx:^10.4f} | "
        f"{z:^12.4f} | {prediction:^12.4f} | "
        f"{predicted_class:^8}"
    )

print("-" * 85)
```
