```python
# ============================================================
# McCulloch-Pitts (MP) Neuron
# Boolean Logic Functions:
# NOT, AND, OR, NAND, NOR, XOR, XNOR
# ============================================================

import itertools


# ============================================================
# McCulloch-Pitts Neuron
# ============================================================

def mp_neuron(inputs, weights, threshold):
    """
    McCulloch-Pitts neuron.

    If weighted sum >= threshold, output = 1
    Otherwise, output = 0
    """

    weighted_sum = 0

    for i in range(len(inputs)):
        weighted_sum += inputs[i] * weights[i]

    if weighted_sum >= threshold:
        return 1
    else:
        return 0


# ============================================================
# NOT Gate
# ============================================================

def NOT(A):
    return mp_neuron(
        inputs=[A],
        weights=[-1],
        threshold=-0.5
    )


# ============================================================
# AND Gate
# ============================================================

def AND(A, B):
    return mp_neuron(
        inputs=[A, B],
        weights=[1, 1],
        threshold=2
    )


# ============================================================
# OR Gate
# ============================================================

def OR(A, B):
    return mp_neuron(
        inputs=[A, B],
        weights=[1, 1],
        threshold=1
    )


# ============================================================
# NAND Gate
# ============================================================

def NAND(A, B):
    return mp_neuron(
        inputs=[A, B],
        weights=[-1, -1],
        threshold=-1.5
    )


# ============================================================
# NOR Gate
# ============================================================

def NOR(A, B):
    return mp_neuron(
        inputs=[A, B],
        weights=[-1, -1],
        threshold=-0.5
    )


# ============================================================
# XOR Gate
# XOR = (A OR B) AND (A NAND B)
# ============================================================

def XOR(A, B):

    # First layer
    or_output = OR(A, B)
    nand_output = NAND(A, B)

    # Output layer
    xor_output = AND(or_output, nand_output)

    return xor_output


# ============================================================
# XNOR Gate
# XNOR = (A AND B) OR (A NOR B)
# ============================================================

def XNOR(A, B):

    # First layer
    and_output = AND(A, B)
    nor_output = NOR(A, B)

    # Output layer
    xnor_output = OR(and_output, nor_output)

    return xnor_output


# ============================================================
# Display NOT Gate
# ============================================================

def display_NOT():

    print("\n")
    print("=" * 44)
    print("NOT GATE")
    print("=" * 44)

    print("Input 1 | Expected Output  | MP Output")
    print("--------|------------------|----------")

    for A in [0, 1]:

        expected_output = 1 - A
        mp_output = NOT(A)

        print(
            f"   {A}    |        "
            f"{expected_output}         |    "
            f"{mp_output}"
        )


# ============================================================
# Display Two-Input Gates
# ============================================================

def display_gate(gate_name, gate_function, expected_outputs):

    print("\n")
    print("=" * 50)
    print(gate_name + " GATE")
    print("=" * 50)

    print("Input 1 | Input 2 | Expected Output  | MP Output")
    print("--------|---------|------------------|----------")

    inputs = list(itertools.product([0, 1], repeat=2))

    for i, (A, B) in enumerate(inputs):

        expected_output = expected_outputs[i]
        mp_output = gate_function(A, B)

        print(
            f"   {A}    |    {B}    |"
            f"        {expected_output}         |"
            f"    {mp_output}"
        )


# ============================================================
# Main Program
# ============================================================

# Display NOT
display_NOT()


# ------------------------------------------------------------
# Expected outputs for each gate
#
# Input combinations are:
# (0,0), (0,1), (1,0), (1,1)
# ------------------------------------------------------------

display_gate(
    "AND",
    AND,
    [0, 0, 0, 1]
)

display_gate(
    "OR",
    OR,
    [0, 1, 1, 1]
)

display_gate(
    "NAND",
    NAND,
    [1, 1, 1, 0]
)

display_gate(
    "NOR",
    NOR,
    [1, 0, 0, 0]
)

display_gate(
    "XOR",
    XOR,
    [0, 1, 1, 0]
)

display_gate(
    "XNOR",
    XNOR,
    [1, 0, 0, 1]
)
```
