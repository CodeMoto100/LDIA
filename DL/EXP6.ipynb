```python
# ============================================================
# CNN FOR HANDWRITTEN DIGIT RECOGNITION
# Dataset: MNIST
# ============================================================

import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt


# ============================================================
# 1. LOAD MNIST DATASET
# ============================================================

(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()


# ============================================================
# 2. PREPROCESSING
# ============================================================

# Normalize pixel values from 0-255 to 0-1
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Add channel dimension
# Original shape: (60000, 28, 28)
# New shape:      (60000, 28, 28, 1)

X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)


print("=" * 70)
print("              CNN - DIGIT RECOGNITION")
print("=" * 70)

print("\nTraining data shape :", X_train.shape)
print("Testing data shape  :", X_test.shape)


# ============================================================
# 3. DESIGN CNN MODEL
# ============================================================

model = models.Sequential([

    # Input Layer
    layers.Input(shape=(28, 28, 1)),

    # --------------------------------------------------------
    # Convolution Layer 1
    # --------------------------------------------------------

    layers.Conv2D(
        filters=32,
        kernel_size=(3, 3),
        activation="relu"
    ),

    # Pooling Layer 1
    layers.MaxPooling2D(
        pool_size=(2, 2)
    ),

    # --------------------------------------------------------
    # Convolution Layer 2
    # --------------------------------------------------------

    layers.Conv2D(
        filters=64,
        kernel_size=(3, 3),
        activation="relu"
    ),

    # Pooling Layer 2
    layers.MaxPooling2D(
        pool_size=(2, 2)
    ),

    # --------------------------------------------------------
    # Flatten
    # --------------------------------------------------------

    layers.Flatten(),

    # --------------------------------------------------------
    # Fully Connected Layer
    # --------------------------------------------------------

    layers.Dense(
        128,
        activation="relu"
    ),

    # --------------------------------------------------------
    # Output Layer
    # 10 neurons = digits 0 to 9
    # --------------------------------------------------------

    layers.Dense(
        10,
        activation="softmax"
    )
])


# ============================================================
# 4. DISPLAY MODEL ARCHITECTURE
# ============================================================

print("\nCNN MODEL ARCHITECTURE")
print("=" * 70)

model.summary()


# ============================================================
# 5. COMPILE MODEL
# ============================================================

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# ============================================================
# 6. TRAIN MODEL
# ============================================================

print("\nTRAINING CNN")
print("=" * 70)

history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.1,
    verbose=1
)


# ============================================================
# 7. EVALUATE MODEL
# ============================================================

print("\nMODEL EVALUATION")
print("=" * 70)

test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print(f"Test Loss     : {test_loss:.4f}")
print(f"Test Accuracy : {test_accuracy * 100:.2f}%")


# ============================================================
# 8. MAKE PREDICTIONS
# ============================================================

predictions = model.predict(
    X_test[:10],
    verbose=0
)

predicted_digits = predictions.argmax(axis=1)


# ============================================================
# 9. DISPLAY PREDICTIONS
# ============================================================

print("\nDIGIT PREDICTIONS")
print("=" * 70)

print(
    f"{'Image':^10} | "
    f"{'Actual':^10} | "
    f"{'Predicted':^10} | "
    f"{'Confidence':^12}"
)

print("-" * 70)

for i in range(10):

    confidence = predictions[i][predicted_digits[i]] * 100

    print(
        f"{i + 1:^10} | "
        f"{y_test[i]:^10} | "
        f"{predicted_digits[i]:^10} | "
        f"{confidence:^11.2f}%"
    )

print("-" * 70)


# ============================================================
# 10. DISPLAY SAMPLE DIGITS
# ============================================================

plt.figure(figsize=(12, 5))

for i in range(10):

    plt.subplot(2, 5, i + 1)

    plt.imshow(
        X_test[i].reshape(28, 28),
        cmap="gray"
    )

    plt.title(
        f"Actual: {y_test[i]}\n"
        f"Pred: {predicted_digits[i]}"
    )

    plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 11. PLOT TRAINING ACCURACY
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("CNN Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()


# ============================================================
# 12. PLOT TRAINING LOSS
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("CNN Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()
```
