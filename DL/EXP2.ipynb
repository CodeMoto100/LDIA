```python
def print_table(title, headers, data):
    print("\n" + title)
    print("-" * 130)
    print(" | ".join(f"{h:^12}" for h in headers))
    print("-" * 130)

    for row in data:
        print(" | ".join(f"{str(x):^12}" for x in row))

    print("-" * 130)


# ---------------- AND GATE ----------------

inputs = [[0,0], [0,1], [1,0], [1,1]]
weights = [1, 1]
bias = -1.5
targets = [0, 0, 0, 1]

data = []

for x, target in zip(inputs, targets):
    wx = x[0]*weights[0] + x[1]*weights[1]
    total = wx + bias
    output = 1 if total >= 0 else 0

    data.append([
        x[0], x[1], weights[0], weights[1],
        wx, bias, total, output, target
    ])

print_table(
    "AND GATE",
    ["X1", "X2", "W1", "W2", "WX", "Bias",
     "WX+B", "MP Output", "Target"],
    data
)


# ---------------- OR GATE ----------------

weights = [1, 1]
bias = -0.5
targets = [0, 1, 1, 1]

data = []

for x, target in zip(inputs, targets):
    wx = x[0]*weights[0] + x[1]*weights[1]
    total = wx + bias
    output = 1 if total >= 0 else 0

    data.append([
        x[0], x[1], weights[0], weights[1],
        wx, bias, total, output, target
    ])

print_table(
    "OR GATE",
    ["X1", "X2", "W1", "W2", "WX", "Bias",
     "WX+B", "MP Output", "Target"],
    data
)


# ---------------- NOT GATE ----------------

inputs = [[0], [1]]
weights = [-1]
bias = 0.5
targets = [1, 0]

data = []

for x, target in zip(inputs, targets):
    wx = x[0] * weights[0]
    total = wx + bias
    output = 1 if total >= 0 else 0

    data.append([
        x[0], weights[0], wx, bias,
        total, output, target
    ])

print_table(
    "NOT GATE",
    ["X", "W", "WX", "Bias", "WX+B",
     "MP Output", "Target"],
    data
)
```
