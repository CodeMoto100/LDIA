```python
# ============================================================
# MULTI-LAYER PERCEPTRON FOR XOR AND XNOR
# ============================================================

# Step activation function
def step(x):
    return 1 if x >= 0 else 0


# ============================================================
# XOR MLP
# ============================================================

def xor_perceptron(x1, x2):

    # -------------------------
    # Hidden Layer
    # -------------------------

    # H1 = OR
    h1_w1 = 1
    h1_w2 = 1
    h1_bias = -0.5

    h1_sum = (x1 * h1_w1) + (x2 * h1_w2)
    h1_total = h1_sum + h1_bias
    h1 = step(h1_total)

    # H2 = NAND
    h2_w1 = -1
    h2_w2 = -1
    h2_bias = 1.5

    h2_sum = (x1 * h2_w1) + (x2 * h2_w2)
    h2_total = h2_sum + h2_bias
    h2 = step(h2_total)

    # -------------------------
    # Output Layer
    # -------------------------

    # Output = AND
    out_w1 = 1
    out_w2 = 1
    out_bias = -1.5

    out_sum = (h1 * out_w1) + (h2 * out_w2)
    out_total = out_sum + out_bias
    output = step(out_total)

    return (
        h1_sum, h1_total, h1,
        h2_sum, h2_total, h2,
        out_sum, out_total, output
    )


# ============================================================
# XNOR MLP
# ============================================================

def xnor_perceptron(x1, x2):

    # -------------------------
    # Hidden Layer
    # -------------------------

    # H1 = AND
    h1_w1 = 1
    h1_w2 = 1
    h1_bias = -1.5

    h1_sum = (x1 * h1_w1) + (x2 * h1_w2)
    h1_total = h1_sum + h1_bias
    h1 = step(h1_total)

    # H2 = NOR
    h2_w1 = -1
    h2_w2 = -1
    h2_bias = 0.5

    h2_sum = (x1 * h2_w1) + (x2 * h2_w2)
    h2_total = h2_sum + h2_bias
    h2 = step(h2_total)

    # -------------------------
    # Output Layer
    # -------------------------

    # Output = OR
    out_w1 = 1
    out_w2 = 1
    out_bias = -0.5

    out_sum = (h1 * out_w1) + (h2 * out_w2)
    out_total = out_sum + out_bias
    output = step(out_total)

    return (
        h1_sum, h1_total, h1,
        h2_sum, h2_total, h2,
        out_sum, out_total, output
    )


# ============================================================
# DATA
# ============================================================

inputs = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

xor_targets = [0, 1, 1, 0]
xnor_targets = [1, 0, 0, 1]


# ============================================================
# EPOCH INFORMATION
# ============================================================

epochs = 1


# ============================================================
# XOR WEIGHTS
# ============================================================

print("\n" + "=" * 120)
print("                         XOR - MULTI-LAYER PERCEPTRON")
print("=" * 120)

print("\nNumber of Epochs :", epochs)

print("\nHidden Layer Weights:")
print("H1 (OR)   : W1 = 1,  W2 = 1,  Bias = -0.5")
print("H2 (NAND) : W1 = -1, W2 = -1, Bias = 1.5")

print("\nFinal Layer Weights:")
print("Output (AND): W1 = 1, W2 = 1, Bias = -1.5")

print("\n" + "-" * 120)

print(
    f"{'X1':^5} | {'X2':^5} | "
    f"{'H1 ΣWX':^8} | {'H1+b':^8} | {'H1':^5} | "
    f"{'H2 ΣWX':^8} | {'H2+b':^8} | {'H2':^5} | "
    f"{'OUT ΣWX':^9} | {'OUT+b':^8} | "
    f"{'Output':^8} | {'Target':^8}"
)

print("-" * 120)

for x, target in zip(inputs, xor_targets):

    result = xor_perceptron(x[0], x[1])

    h1_sum, h1_total, h1, \
    h2_sum, h2_total, h2, \
    out_sum, out_total, output = result

    print(
        f"{x[0]:^5} | {x[1]:^5} | "
        f"{h1_sum:^8} | {h1_total:^8} | {h1:^5} | "
        f"{h2_sum:^8} | {h2_total:^8} | {h2:^5} | "
        f"{out_sum:^9} | {out_total:^8} | "
        f"{output:^8} | {target:^8}"
    )

print("-" * 120)


# ============================================================
# XNOR WEIGHTS
# ============================================================

print("\n" + "=" * 120)
print("                        XNOR - MULTI-LAYER PERCEPTRON")
print("=" * 120)

print("\nNumber of Epochs :", epochs)

print("\nHidden Layer Weights:")
print("H1 (AND) : W1 = 1,  W2 = 1,  Bias = -1.5")
print("H2 (NOR) : W1 = -1, W2 = -1, Bias = 0.5")

print("\nFinal Layer Weights:")
print("Output (OR): W1 = 1, W2 = 1, Bias = -0.5")

print("\n" + "-" * 120)

print(
    f"{'X1':^5} | {'X2':^5} | "
    f"{'H1 ΣWX':^8} | {'H1+b':^8} | {'H1':^5} | "
    f"{'H2 ΣWX':^8} | {'H2+b':^8} | {'H2':^5} | "
    f"{'OUT ΣWX':^9} | {'OUT+b':^8} | "
    f"{'Output':^8} | {'Target':^8}"
)

print("-" * 120)

for x, target in zip(inputs, xnor_targets):

    result = xnor_perceptron(x[0], x[1])

    h1_sum, h1_total, h1, \
    h2_sum, h2_total, h2, \
    out_sum, out_total, output = result

    print(
        f"{x[0]:^5} | {x[1]:^5} | "
        f"{h1_sum:^8} | {h1_total:^8} | {h1:^5} | "
        f"{h2_sum:^8} | {h2_total:^8} | {h2:^5} | "
        f"{out_sum:^9} | {out_total:^8} | "
        f"{output:^8} | {target:^8}"
    )

print("-" * 120)
```
