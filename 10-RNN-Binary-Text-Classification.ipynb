```python
# =====================================================================
# LSTM BASED RECURRENT NEURAL NETWORK FOR BINARY TEXT CLASSIFICATION
# SPAM vs HAM SMS MESSAGES
# Google Colab - COMPLETE SINGLE CELL
# =====================================================================

# =====================================================================
# 1. IMPORT LIBRARIES
# =====================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
from sklearn.utils.class_weight import compute_class_weight

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

from tensorflow.keras import Sequential
from tensorflow.keras.layers import (
    Input,
    Embedding,
    Bidirectional,
    LSTM,
    Dense,
    Dropout
)

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)

print("TensorFlow Version:", tf.__version__)
print("GPU Available:", tf.config.list_physical_devices("GPU"))


# =====================================================================
# 2. LOAD SMS SPAM DATASET
# =====================================================================

url = "https://raw.githubusercontent.com/justmarkham/pycon-2016-tutorial/master/data/sms.tsv"

df = pd.read_csv(
    url,
    sep="\t",
    header=None,
    names=["label", "message"]
)

print("\n================ DATASET ================")
print("Dataset shape:", df.shape)

print("\nFirst 5 records:")
print(df.head())


# =====================================================================
# 3. CHECK DATA
# =====================================================================

print("\n================ CLASS DISTRIBUTION ================")
print(df["label"].value_counts())

print("\nMissing values:")
print(df.isnull().sum())


# =====================================================================
# 4. CONVERT LABELS
# =====================================================================
# Ham  = 0
# Spam = 1

df["label"] = df["label"].map({
    "ham": 0,
    "spam": 1
})

print("\nEncoded labels:")
print(df["label"].value_counts())


# =====================================================================
# 5. DISPLAY CLASS DISTRIBUTION
# =====================================================================

plt.figure(figsize=(6, 4))

sns.countplot(
    x=df["label"],
    hue=df["label"],
    palette=["green", "red"],
    legend=False
)

plt.xticks(
    [0, 1],
    ["Ham", "Spam"]
)

plt.xlabel("Message Type")
plt.ylabel("Number of Messages")
plt.title("Ham vs Spam Messages")

plt.show()


# =====================================================================
# 6. SEPARATE INPUT AND OUTPUT
# =====================================================================

X = df["message"].values
y = df["label"].values


# =====================================================================
# 7. TRAIN / TEST SPLIT
# =====================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n================ DATA SPLIT ================")
print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))


# =====================================================================
# 8. TEXT TOKENIZATION
# =====================================================================

vocab_size = 10000
max_length = 100

tokenizer = Tokenizer(
    num_words=vocab_size,
    oov_token="<OOV>"
)

# Fit tokenizer ONLY on training data
tokenizer.fit_on_texts(X_train)

X_train_sequences = tokenizer.texts_to_sequences(X_train)
X_test_sequences = tokenizer.texts_to_sequences(X_test)


# =====================================================================
# 9. PAD SEQUENCES
# =====================================================================

X_train_padded = pad_sequences(
    X_train_sequences,
    maxlen=max_length,
    padding="post",
    truncating="post"
)

X_test_padded = pad_sequences(
    X_test_sequences,
    maxlen=max_length,
    padding="post",
    truncating="post"
)

print("\n================ TEXT PROCESSING ================")
print("Vocabulary size:", len(tokenizer.word_index))
print("Training input :", X_train_padded.shape)
print("Testing input  :", X_test_padded.shape)


# =====================================================================
# 10. HANDLE CLASS IMBALANCE
# =====================================================================
#
# The original model learned:
#
#        EVERYTHING -> HAM
#
# because Ham is much more common than Spam.
#
# Class weights force the model to pay more attention to Spam.
# =====================================================================

classes = np.unique(y_train)

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train
)

class_weights = dict(
    zip(classes, weights)
)

print("\n================ CLASS WEIGHTS ================")

print("Ham weight :", class_weights[0])
print("Spam weight:", class_weights[1])


# =====================================================================
# 11. BUILD LSTM MODEL
# =====================================================================

model = Sequential([

    # Explicit input shape removes the "unbuilt model" issue
    Input(shape=(max_length,)),

    # Convert word IDs into dense vectors
    Embedding(
        input_dim=vocab_size,
        output_dim=128
    ),

    # Bidirectional LSTM
    Bidirectional(
        LSTM(
            64,
            return_sequences=False
        )
    ),

    Dropout(0.5),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(0.3),

    # Binary output
    Dense(
        1,
        activation="sigmoid"
    )
])


# =====================================================================
# 12. MODEL SUMMARY
# =====================================================================

print("\n================ MODEL SUMMARY ================")

model.summary()


# =====================================================================
# 13. COMPILE MODEL
# =====================================================================

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0005
    ),

    loss="binary_crossentropy",

    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall")
    ]
)


# =====================================================================
# 14. CALLBACKS
# =====================================================================

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=4,
    restore_best_weights=True
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=2,
    min_lr=1e-6,
    verbose=1
)


# =====================================================================
# 15. TRAIN MODEL
# =====================================================================

print("\n================ TRAINING ================")

history = model.fit(

    X_train_padded,

    y_train,

    validation_split=0.20,

    epochs=15,

    batch_size=32,

    shuffle=True,

    # IMPORTANT:
    # Gives Spam a higher training weight
    class_weight=class_weights,

    callbacks=[
        early_stopping,
        reduce_lr
    ],

    verbose=1
)


# =====================================================================
# 16. PLOT TRAINING HISTORY
# =====================================================================

plt.figure(figsize=(15, 5))


# Accuracy
plt.subplot(1, 3, 1)

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Training vs Validation Accuracy")
plt.legend()
plt.grid(alpha=0.3)


# Loss
plt.subplot(1, 3, 2)

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training vs Validation Loss")
plt.legend()
plt.grid(alpha=0.3)


# Recall
plt.subplot(1, 3, 3)

plt.plot(
    history.history["recall"],
    label="Training Recall"
)

plt.plot(
    history.history["val_recall"],
    label="Validation Recall"
)

plt.xlabel("Epoch")
plt.ylabel("Recall")
plt.title("Spam Recall")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()


# =====================================================================
# 17. TEST MODEL
# =====================================================================

print("\n================ TEST RESULTS ================")

test_results = model.evaluate(
    X_test_padded,
    y_test,
    verbose=0
)

print(
    f"Test Loss      : {test_results[0]:.4f}"
)

print(
    f"Test Accuracy  : {test_results[1] * 100:.2f}%"
)

print(
    f"Test Precision : {test_results[2] * 100:.2f}%"
)

print(
    f"Test Recall    : {test_results[3] * 100:.2f}%"
)


# =====================================================================
# 18. PREDICT TEST DATA
# =====================================================================

y_probability = model.predict(
    X_test_padded,
    verbose=0
).flatten()


# Default threshold = 0.5
y_pred = (
    y_probability >= 0.5
).astype(int)


# =====================================================================
# 19. CALCULATE PERFORMANCE METRICS
# =====================================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)


print("\n================ PERFORMANCE METRICS ================")

print(
    f"Accuracy  : {accuracy * 100:.2f}%"
)

print(
    f"Precision : {precision * 100:.2f}%"
)

print(
    f"Recall    : {recall * 100:.2f}%"
)

print(
    f"F1-Score  : {f1 * 100:.2f}%"
)


# =====================================================================
# 20. CLASSIFICATION REPORT
# =====================================================================

print("\n================ CLASSIFICATION REPORT ================")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Ham", "Spam"],
        zero_division=0
    )
)


# =====================================================================
# 21. CONFUSION MATRIX
# =====================================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\n================ CONFUSION MATRIX ================")

print(cm)


plt.figure(figsize=(7, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Ham", "Spam"],
    yticklabels=["Ham", "Spam"]
)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("Confusion Matrix")

plt.show()


# =====================================================================
# 22. CHECK PREDICTION DISTRIBUTION
# =====================================================================

print("\n================ PREDICTION DISTRIBUTION ================")

unique, counts = np.unique(
    y_pred,
    return_counts=True
)

for label, count in zip(unique, counts):

    if label == 0:
        print("Predicted Ham :", count)

    else:
        print("Predicted Spam:", count)


# =====================================================================
# 23. TEST CUSTOM SMS MESSAGES
# =====================================================================

custom_messages = [

    "Congratulations! You have won a free prize. Click now to claim your reward.",

    "Hey, are we meeting for lunch today?",

    "URGENT! You have won 1000 dollars. Call now to claim your prize.",

    "Can you send me the notes from today's class?",

    "You have been selected for a special cash reward. Reply WIN now.",

    "I will reach home by 7 pm. See you soon.",

    "FREE entry into our cash prize competition. Text WIN to 80085 now.",

    "Your appointment is confirmed for tomorrow at 10 AM."
]


# Convert text to sequences
custom_sequences = tokenizer.texts_to_sequences(
    custom_messages
)

custom_padded = pad_sequences(
    custom_sequences,
    maxlen=max_length,
    padding="post",
    truncating="post"
)


# Predict probabilities
custom_probabilities = model.predict(
    custom_padded,
    verbose=0
).flatten()


print("\n================ CUSTOM MESSAGE PREDICTIONS ================")

for message, probability in zip(
    custom_messages,
    custom_probabilities
):

    if probability >= 0.5:
        prediction = "SPAM"
    else:
        prediction = "HAM"

    print("\nMessage:")
    print(message)

    print(
        f"Prediction      : {prediction}"
    )

    print(
        f"Spam Probability: {probability * 100:.2f}%"
    )


# =====================================================================
# 24. SHOW SOME TEST PREDICTIONS
# =====================================================================

print("\n================ SAMPLE TEST PREDICTIONS ================")

for i in range(20):

    actual = (
        "SPAM"
        if y_test[i] == 1
        else "HAM"
    )

    predicted = (
        "SPAM"
        if y_pred[i] == 1
        else "HAM"
    )

    print(
        f"{i+1:2d}. Actual: {actual:4s} | "
        f"Predicted: {predicted:4s} | "
        f"Spam Probability: {y_probability[i]*100:.2f}%"
    )


# =====================================================================
# 25. SAVE MODEL
# =====================================================================

model.save(
    "spam_ham_lstm.keras"
)

import pickle

with open(
    "spam_ham_tokenizer.pkl",
    "wb"
) as f:

    pickle.dump(
        tokenizer,
        f
    )


# =====================================================================
# 26. FINAL SUMMARY
# =====================================================================

print("\n")
print("=" * 65)
print("          FINAL LSTM SPAM/HAM RESULTS")
print("=" * 65)

print(
    f"Vocabulary Size       : {vocab_size}"
)

print(
    f"Maximum Sequence Length: {max_length}"
)

print(
    f"Embedding Dimension   : 128"
)

print(
    f"LSTM Units            : 64"
)

print(
    f"Model Type            : Bidirectional LSTM"
)

print(
    f"Test Accuracy         : {accuracy * 100:.2f}%"
)

print(
    f"Precision             : {precision * 100:.2f}%"
)

print(
    f"Recall                : {recall * 100:.2f}%"
)

print(
    f"F1-Score              : {f1 * 100:.2f}%"
)

print("=" * 65)

print("\nModel saved as:")
print("spam_ham_lstm.keras")

print("\nTokenizer saved as:")
print("spam_ham_tokenizer.pkl")

print("\nExperiment completed successfully!")
```
